```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

sns.set_theme(style="whitegrid")
df = pd.read_csv("../data/student_performance_simulated.csv")
df.head()
```

## 2. Pass / Fail Analysis

## 5. Hypothesis Test

### Question
Do students with attendance of at least 85% have a different average final score from students with attendance below 70%?

- **H0:** The two groups have equal mean final scores.
- **H1:** The two groups have different mean final scores.

A Welch independent-samples t-test is used because the groups may have different variances.


```python
high = df.loc[df["Attendance"] >= 85, "Final_Score"]
low = df.loc[df["Attendance"] < 70, "Final_Score"]

t_stat, p_value = stats.ttest_ind(high, low, equal_var=False)

print("High-attendance group mean:", round(high.mean(),2))
print("Low-attendance group mean:", round(low.mean(),2))
print("t-statistic:", round(t_stat,3))
print("p-value:", p_value)
```

### Interpretation

If the p-value is below the selected significance level (commonly 0.05), the sample provides statistical evidence of a difference between the two group means.

This test identifies an association between attendance group and scores; it does **not** prove that attendance itself causes the difference.
