# Project 5 — Finance Market Analysis

**Dataset note:** This project uses simulated market and sentiment data for portfolio practice. It is not real market data and is not investment advice.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv('../data/finance_market_data_simulated_raw.csv',parse_dates=['Date'])
df.head()
```

## 1. Data Cleaning
- Inspect missing values and duplicates.
- Remove duplicate records.
- Median-impute selected numeric fields.
- Create year, month and positive-return features.

```python
print(df.shape)
print(df.isna().sum())
print('Duplicates:',df.duplicated().sum())
clean=df.drop_duplicates().copy()
for c in ['Sentiment_Score','Trading_Volume']: clean[c]=clean[c].fillna(clean[c].median())
clean['Year']=clean.Date.dt.year
clean['Month']=clean.Date.dt.to_period('M').astype(str)
clean['Positive_Return']=clean.Daily_Return>0
clean.describe()
```

## 2. Statistical Analysis
Annualized volatility is estimated as daily return standard deviation multiplied by √252. Correlation describes association, not causation.

```python
sector=clean.groupby('Asset_Sector').agg(Avg_Return=('Daily_Return',lambda s:s.mean()*100),Volatility=('Daily_Return',lambda s:s.std()*np.sqrt(252)*100),Avg_Volume=('Trading_Volume','mean'),Avg_Sentiment=('Sentiment_Score','mean'))
sector
```

```python
corr=clean[['Daily_Return','Trading_Volume','Sentiment_Score','Engagement_Index']].corr()
corr
```

## 3. Visualizations
Six charts are exported to the `visualizations/` folder.

```python
monthly=clean.groupby('Month').Daily_Return.mean()*100
monthly.plot(title='Average Daily Return by Month'); plt.ylabel('Return (%)'); plt.xticks(rotation=45); plt.show()
```

```python
plt.scatter(clean.Sentiment_Score,clean.Daily_Return,alpha=.25)
plt.title('Sentiment Score vs Daily Return'); plt.xlabel('Sentiment'); plt.ylabel('Daily Return'); plt.show()
```

## 4. Key Findings
- The dataset contains 1,500 simulated observations after cleaning.
- Sector and market groupings provide examples of comparative financial analysis.
- Volatility measures describe dispersion of returns and should not be treated as a forecast.
- Sentiment-return correlation is observational and does not establish causation.
