# Project 3 — Weather Data Analysis

**Dataset type:** Simulated daily weather dataset for portfolio practice.

## Objectives
- Analyze temperature trends
- Identify seasonal patterns
- Analyze rainfall distribution
- Examine temperature/humidity relationships
- Identify extreme weather observations
- Apply descriptive statistics and correlation analysis


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
df=pd.read_csv("../data/weather_simulated.csv",parse_dates=["Date"])
df.head()
```

## 1. Data Quality Validation

```python
print("Shape:",df.shape)
print("Missing cells:",df.isna().sum().sum())
print("Duplicate rows:",df.duplicated().sum())
print(df.describe())
```

```python
assert df["Rainfall_mm"].ge(0).all()
assert df["Humidity_pct"].between(0,100).all()
assert df["Date"].notna().all()
print("Validation checks passed.")
```

## 2. Temperature Trend

```python
df["30D_Temp_Avg"]=df["Temperature_C"].rolling(30).mean()
plt.figure(figsize=(10,5))
plt.plot(df["Date"],df["Temperature_C"],alpha=.5)
plt.plot(df["Date"],df["30D_Temp_Avg"],linewidth=2,label="30-day average")
plt.title("Daily Temperature and 30-Day Rolling Average")
plt.xlabel("Date"); plt.ylabel("Temperature (°C)"); plt.legend()
plt.show()
```

## 3. Seasonal and Monthly Analysis

```python
monthly=df.groupby("Month").agg(
    Avg_Temperature=("Temperature_C","mean"),
    Total_Rainfall=("Rainfall_mm","sum"),
    Avg_Humidity=("Humidity_pct","mean")
)
monthly
```

```python
monthly["Total_Rainfall"].plot(kind="bar",figsize=(9,5),title="Total Rainfall by Month")
plt.xlabel("Month"); plt.ylabel("Rainfall (mm)")
plt.show()
```

## 4. Rainfall Distribution

```python
print("Rainy days:",(df["Rainfall_mm"]>0).sum())
print("Maximum daily rainfall:",df["Rainfall_mm"].max())
plt.figure(figsize=(9,5))
plt.hist(df["Rainfall_mm"],bins=30)
plt.title("Daily Rainfall Distribution")
plt.xlabel("Rainfall (mm)"); plt.ylabel("Days")
plt.show()
```

## 5. Correlation Analysis

```python
corr=df[["Temperature_C","Humidity_pct","Rainfall_mm","Wind_Speed_kmh","Pressure_hPa"]].corr()
corr
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(corr,annot=True,fmt=".2f")
plt.title("Weather Variable Correlations")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Temperature_C"],df["Humidity_pct"],alpha=.25)
plt.title("Temperature vs Humidity")
plt.xlabel("Temperature (°C)"); plt.ylabel("Humidity (%)")
plt.show()
```

## 6. Extreme Weather

```python
low=df["Temperature_C"].quantile(.02)
high=df["Temperature_C"].quantile(.98)
extremes=df[(df["Temperature_C"]<low)|(df["Temperature_C"]>high)]
print("2nd percentile:",low)
print("98th percentile:",high)
print("Extreme-temperature days:",len(extremes))
print("\nHottest day:")
display(df.loc[[df["Temperature_C"].idxmax()]])
print("\nWettest day:")
display(df.loc[[df["Rainfall_mm"].idxmax()]])
```

## 7. Key Findings

1. Temperature follows a recurring seasonal pattern across the simulated three-year period.
2. Rainfall is unevenly distributed across months, with some months contributing much more total rainfall.
3. Daily rainfall is right-skewed: many days have little/no rain while a smaller number of days have heavier rainfall.
4. Temperature and humidity show a negative association in the simulated data.
5. Extreme-temperature observations can be identified using percentile thresholds.
6. Rolling averages make longer-term temperature patterns easier to interpret than raw daily fluctuations.


## 8. Recommendations

1. Use monthly and seasonal weather summaries for operational planning.
2. Monitor rainfall extremes separately from average rainfall because averages can hide heavy-rain events.
3. Use rolling temperature measures to identify sustained changes rather than reacting to individual days.
4. For real-world forecasting or climate conclusions, combine this type of analysis with longer, validated observational datasets.


## 9. Limitations

- The dataset is simulated.
- It represents one modeled weather series rather than an official meteorological station.
- Correlation does not establish causation.
- Extreme-event thresholds are analytical definitions, not official weather warnings.
