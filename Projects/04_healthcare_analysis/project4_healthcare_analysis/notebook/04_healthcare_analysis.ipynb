# Project 4 — Healthcare Patient Analysis

**Dataset:** Simulated healthcare data for portfolio practice. It is not real patient data and must not be used for clinical decisions.

```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('../data/healthcare_patient_data_simulated_raw.csv')
df.head()
```

## 1. Data Cleaning and Validation
- Inspect missing values and duplicates.
- Remove exact duplicate rows.
- Median-impute selected numeric fields.

```python
print(df.shape)
print(df.isna().sum())
print('Duplicates:',df.duplicated().sum())
clean=df.drop_duplicates().copy()
for c in ['BMI','Cholesterol','Glucose']: clean[c]=clean[c].fillna(clean[c].median())
clean.describe()
```

## 2. Statistical Analysis

```python
summary=clean.groupby('Condition').agg(Patients=('Patient_ID','count'),Avg_Age=('Age','mean'),Avg_Visits=('Annual_Visits','mean'),High_Risk_Rate=('High_Risk',lambda s:(s=='Yes').mean()*100))
summary.sort_values('Patients',ascending=False)
```

```python
corr=clean[['Age','BMI','Systolic_BP','Cholesterol','Glucose','Annual_Visits']].corr()
corr
```

## 3. Visualizations
The exported `visualizations/` folder contains six publication-ready charts corresponding to the analysis.

```python
clean['Condition'].value_counts().plot(kind='barh',title='Patient Distribution by Condition'); plt.xlabel('Patients'); plt.show()
```

```python
agegrp=pd.cut(clean.Age,[17,29,39,49,59,69,79,100],labels=['18–29','30–39','40–49','50–59','60–69','70–79','80+'])
r=clean.assign(Age_Group=agegrp).groupby('Age_Group',observed=False).High_Risk.apply(lambda s:(s=='Yes').mean()*100)
r.plot(kind='bar',title='High-Risk Rate by Age Group'); plt.ylabel('High-Risk %'); plt.show()
```

## 4. Key Findings
- Cleaned records: 1,000.
- Average age: 51.7 years.
- Average BMI: 26.7.
- Average annual visits: 4.35.
- Synthetic high-risk rate: 30.2%.

## 5. Limitations
All records and risk labels are synthetic. Correlation does not imply causation, and these results have no clinical validity.
